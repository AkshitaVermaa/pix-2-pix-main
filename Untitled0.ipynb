```python
import os, time, torch
from torch.utils.data import DataLoader
from model import PairedImageDataset, Pix2PixModel, save_sample_images, plot_training_curves

CKPT_DIR = "/content/drive/MyDrive/pix2pix_checkpoints"
os.makedirs(CKPT_DIR, exist_ok=True)
os.makedirs("results", exist_ok=True)

dataset = PairedImageDataset("/content/work/dataset", image_size=256)   # expect 570 pairs, 114 subjects
loader  = DataLoader(dataset, batch_size=1, shuffle=True, num_workers=2, pin_memory=True)

model = Pix2PixModel(lr=0.0002)
start = model.load_checkpoint("/content/work/checkpoints/pix2pix_final.pth")  # returns 50
TOTAL = 150

history = {k: [] for k in ["loss_D","loss_G","loss_G_GAN","loss_G_L1","loss_G_L2"]}
for epoch in range(start, TOTAL):
    t0 = time.time()
    ep = {k: 0 for k in history}
    for side, front in loader:
        l = model.train_step(side, front)
        for k in ep: ep[k] += l[k]
    for k in ep:
        ep[k] /= len(loader); history[k].append(ep[k])
    print(f"Epoch [{epoch+1}/{TOTAL}] L1: {ep['loss_G_L1']:.4f} L2: {ep['loss_G_L2']:.4f} "
          f"BCE: {ep['loss_G_GAN']:.4f} D: {ep['loss_D']:.4f} Time: {time.time()-t0:.1f}s")
    if (epoch + 1) % 10 == 0:
        model.save_checkpoint(f"{CKPT_DIR}/pix2pix_epoch_{epoch+1}.pth", epoch+1)
        model.save_checkpoint(f"{CKPT_DIR}/pix2pix_last.pth", epoch+1)   # always the newest
        s, f_ = next(iter(loader))
        with torch.no_grad():
            fake = model.generator(s.to(model.device))
        save_sample_images(s.to(model.device), f_, fake, epoch+1)

model.save_checkpoint(f"{CKPT_DIR}/pix2pix_final_150.pth", TOTAL)
plot_training_curves(history)
```

```python
import torch, os
D = "/content/drive/MyDrive/pix2pix_checkpoints"
final = torch.load(f"{D}/pix2pix_final_150.pth", map_location="cpu")["generator_state_dict"]

def params_only(sd):
    return {k: v for k, v in sd.items() if "running" not in k and "num_batches" not in k}

fp = params_only(final)
for name in ["pix2pix_epoch_150.pth", "pix2pix_last.pth"]:
    o = torch.load(f"{D}/{name}", map_location="cpu")["generator_state_dict"]
    op = params_only(o)
    same = all(torch.equal(op[k], fp[k]) for k in fp)
    maxdiff = max((op[k].float() - fp[k].float()).abs().max().item() for k in fp)
    print(f"{name}: learned weights identical = {same}, max difference = {maxdiff}")
    del o
```

```python
import os, torch, zipfile, shutil
from PIL import Image
from torchvision import transforms
from model import UNetGenerator
import matplotlib.pyplot as plt

device = "cuda" if torch.cuda.is_available() else "cpu"
CKPT = "/content/drive/MyDrive/pix2pix_checkpoints/pix2pix_final_150.pth"
SRC  = "/content/work/dataset"
OUT  = "/content/frontalized_all"          # local, fast
ZIP  = "/content/drive/MyDrive/frontalized_all.zip"

# --- load generator ---
G = UNetGenerator().to(device)
G.load_state_dict(torch.load(CKPT, map_location="cpu")["generator_state_dict"])
G.train()            # BatchNorm uses batch statistics (best with batch size 1); set G.eval() to compare
torch.manual_seed(0) # dropout is active in train mode, so fix the seed for repeatable results

tf = transforms.Compose([transforms.Resize((256, 256)), transforms.ToTensor(),
                         transforms.Normalize([0.5]*3, [0.5]*3)])

# --- generate all subjects ---
shutil.rmtree(OUT, ignore_errors=True)
count, subjects = 0, sorted(d for d in os.listdir(SRC) if d.startswith("subject_"))
with torch.no_grad():
    for s in subjects:
        os.makedirs(f"{OUT}/{s}", exist_ok=True)
        for f in sorted(os.listdir(f"{SRC}/{s}")):
            if not f.startswith("side"):
                continue
            x = tf(Image.open(f"{SRC}/{s}/{f}").convert("RGB")).unsqueeze(0).to(device)
            y = (G(x)[0].cpu() * 0.5 + 0.5).clamp(0, 1)
            transforms.ToPILImage()(y).save(f"{OUT}/{s}/frontal_{f}", quality=95)
            count += 1
print(f"Subjects: {len(subjects)} | Generated: {count} images (expect 114 and 570)")

# --- preview: all 5 sides for 3 subjects (side | generated | real front) ---
for s in [subjects[0], subjects[56], subjects[-1]]:
    fig, ax = plt.subplots(5, 3, figsize=(6, 10))
    for i in range(5):
        ax[i,0].imshow(Image.open(f"{SRC}/{s}/side{i+1}.jpg"))
        ax[i,1].imshow(Image.open(f"{OUT}/{s}/frontal_side{i+1}.jpg"))
        ax[i,2].imshow(Image.open(f"{SRC}/{s}/front.jpg"))
        for a in ax[i]: a.axis("off")
    ax[0,0].set_title("Side"); ax[0,1].set_title("Generated"); ax[0,2].set_title("Real front")
    fig.suptitle(s); plt.tight_layout(); plt.show()

# --- zip to Drive ---
with zipfile.ZipFile(ZIP, "w", zipfile.ZIP_DEFLATED) as z:
    for root, _, files in os.walk(OUT):
        for f in files:
            p = os.path.join(root, f)
            z.write(p, os.path.relpath(p, "/content"))   # keeps frontalized_all/subject_XXX/... inside the zip
print("Saved zip:", ZIP, f"({os.path.getsize(ZIP)/1e6:.1f} MB)")
```
